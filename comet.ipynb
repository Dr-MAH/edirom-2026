```python
from datetime import datetime
from datetime import timedelta

feb_8_1986: datetime = datetime(1986, 2, 8)
jul_28_2061: datetime = datetime(2061, 7, 28)

num_days_between: int = (jul_28_2061 - feb_8_1986).days
print(num_days_between)

for i in range(1, 6):
    next_year: datetime = feb_8_1986 + timedelta(days=num_days_between * i)
    print(f"Halley's comet will next appear {next_year.strftime("%Y-%m-%d")}")

for i in range(1, 21):
    last_year: datetime = feb_8_1986 - timedelta(days=num_days_between * i)
    print(f"Halley's comet last appeared {last_year.strftime("%Y-%m-%d")}")
```

# Negative Dates

The Python datetime module crashes when it tries to work with dates less than 1 CE. To get around this we need external modules to handle computing dates. 

There are several, with varying degrees of intelligence about calendars and leap years. For this example we will use NumPy's datetime64 module. 

```python
import numpy as np

for i in range(1, 31):
    np_date = np.datetime64(feb_8_1986) - np.timedelta64(num_days_between * i, 'D')
    print(f"Halley's comet last appeared {np.datetime_as_string(np_date, unit='D')}")
```
